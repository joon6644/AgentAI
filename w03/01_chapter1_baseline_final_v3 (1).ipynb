## 🛠️ 프로젝트 개발 환경 구축

이 노트북은 **CrewAI**와 **OpenAI**를 활용한 에이전트 개발 환경 설정을 안내합니다.  
아래 단계에 따라 가상환경을 구축하고 필요한 패키지를 설치해 주세요.

터미널(Terminal)에서(Ctrl+Shift+`) 실행합니다.

* **가상환경 생성:** `env01`이라는 독립된 환경을 구축합니다.

* **가상환경 활성화:** 생성된 환경으로 진입합니다. (Windows CMD 기준)
    ```bash
    env01\Scripts\activate.bat 
    ```

### 2. 필수 라이브러리 설치
진입한 가상환경 터미널에서 아래 명령어를 실행하여 핵심 패키지들을 설치합니다.

| 패키지 | 용도 |
| :--- | :--- |
| **`crewai`** | Agentic AI 기반 멀티 에이전트 협업 프레임워크 |
| **`openai`** | GPT 모델 연동 API |
| **`python-dotenv`** | 환경 변수(.env) 로드 |


```bash
pip install crewai openai python-dotenv
```

### 1️⃣ 환경 변수 로드

````python
# 아래의 코드는 .env 파일에서 OPENAI_MODEL 환경 변수를 로드하는 예시입니다.
# .env 파일에 OPENAI_MODEL이 정의되어 있어야 합니다.
# 또한, .env 파일에는 OPENAI_API_KEY도 정의되어 있어야 합니다.

import os
from dotenv import load_dotenv

load_dotenv() # .env 파일에서 환경 변수를 로드합니다.

OPENAI_MODEL = os.getenv("OPENAI_MODEL") #

print(f"OPENAI_MODEL: {OPENAI_MODEL}")  # OPENAI_MODEL이 제대로 로드되었는지 확인하기 위해 출력합니다.
````

### 2️⃣ Agent 정의 (Tool 없음)

````python
from crewai import Agent, LLM

llm = LLM(model=OPENAI_MODEL)

baseline_agent = Agent(
    role="AI 기반 데이터 분석 전략 컨설턴트",
    goal=(
        "최근 3년 이내 기업들의 AI 기반 데이터 분석 자동화 도입 사례를 조사하고 "
        "근거 기반으로 정리한다."
    ),
    backstory=(
        "기업 디지털 전환 전략을 자문하는 전문가로서 "
        "출처 기반 리서치를 중시한다. "
        "단, 이번 실습에서는 외부 검색 없이 학습된 지식만으로 사례를 작성한다."
    ),
    verbose=True,           # 상세한 로그 출력
    allow_delegation=False, # Agent가 작업을 다른 Agent에게 위임하는 것을 허용하지 않음
    llm=llm
)
````

### 3️⃣ Task 정의

````python
from crewai import Task

baseline_task = Task(
    description=(
        "최근 3년 이내 기업들의 AI 기반 데이터 분석 자동화 도입 사례를 최소 3개 조사하라. "
        "도입 배경, 실행 방식, 성과를 정리하라."
    ),
    expected_output=(
        "각 사례별로:\n"
        "1. 기업명\n"
        "2. 도입 배경\n"
        "3. 실행 방식\n"
        "4. 성과\n"
        "5. 출처(URL 포함, 불확실하면 '출처 미확인'으로 표기)\n"
        "최소 3개 사례 포함."
    ),
    agent=baseline_agent # 이 Task를 baseline_agent에게 할당
)
````

### 4️⃣ Crew 구성

````python
from crewai import Crew

baseline_crew = Crew(
    agents=[baseline_agent],
    tasks=[baseline_task],
    verbose=True
)
````

### 5️⃣ Crew 실행과 최종 결과 출력

````python
result = baseline_crew.kickoff()

print("\n===== 최종 결과 =====\n")
print(result)
````

### 6️⃣ 결과 저장

````python
from datetime import datetime

# 문자열로 안전하게 변환
output_text = str(result)

# 존재하지 않으면 폴더 생성
os.makedirs("outputs", exist_ok=True)

# 파일명 생성
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
filename = f"outputs/chapter1_baseline_{timestamp}.md"

# 파일 저장
with open(filename, "w", encoding="utf-8") as f:
    f.write("# 1장 실습 결과 (Baseline)\n\n")
    f.write(output_text)

print(f"\n✅ 결과가 다음 파일에 저장되었습니다:\n{filename}")
````
